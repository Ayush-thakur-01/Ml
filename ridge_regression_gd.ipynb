```python
import numpy as np

from sklearn.datasets import load_diabetes
from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score
```

```python
X, y = load_diabetes(return_X_y=True)

print(X.shape)
print(y.shape)
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.2,
    random_state=2
)
```

```python
X_train = np.insert(
    X_train,
    0,
    1,
    axis=1
)

X_test = np.insert(
    X_test,
    0,
    1,
    axis=1
)
```

```python
class RidgeGD:

    def __init__(self, alpha=0.1, learning_rate=0.01, epochs=1000):

        self.alpha = alpha
        self.learning_rate = learning_rate
        self.epochs = epochs

    def fit(self, X, y):

        self.coef_ = np.zeros(X.shape[1])

        for i in range(self.epochs):

            y_pred = np.dot(X, self.coef_)

            error = y_pred - y

            gradient = (
                (2 / X.shape[0])
                * np.dot(X.T, error)
            )

            # Ridge regularization
            reg = 2 * self.alpha * self.coef_

            # Don't regularize intercept
            reg[0] = 0

            gradient = gradient + reg

            # Update coefficients
            self.coef_ = (
                self.coef_
                - self.learning_rate * gradient
            )

    def predict(self, X):

        return np.dot(X, self.coef_)
```

```python
reg = RidgeGD(
    alpha=0.1,
    learning_rate=0.01,
    epochs=1000
)
```

```python
reg.fit(X_train, y_train)
```

```python
print(reg.coef_)
```

```python
y_pred = reg.predict(X_test)
```

```python
from sklearn.metrics import r2_score

print(r2_score(y_test, y_pred))
```

```python
for alpha in [0, 0.01, 0.1, 1, 10, 100]:

    reg = RidgeGD(
        alpha=alpha,
        learning_rate=0.01,
        epochs=1000
    )

    reg.fit(X_train, y_train)

    y_pred = reg.predict(X_test)

    print(
        "Alpha:", alpha,
        "R2:", r2_score(y_test, y_pred)
    )
```
